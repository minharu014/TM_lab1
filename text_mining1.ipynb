Loop to remove duplicates from list (Pandas library has a special function for this when dealing with data)

```python
a = [10,20,30,20,10,50,60,40,80,50,40]
dup_items = set()                   # stores unique items
uniq_items = []                     # may contain duplicates but contains the same as the set in this example but in different order

for x in a:                         # List comprehension is a more efficient way to do something similar
    if x not in dup_items:
        uniq_items.append(x)
        dup_items.add(x)

print(uniq_items)
print(dup_items)
```

Conditional statements

```python
n=6
c=0
odds, even = [], []
elem = [3,7,6,77,21,10]

for i in range(0,n):
   if elem[i]%2 == 1:           # i.e. Odd numbers
       odds.append(elem[i])
       c+=1
   elif elem[i]%2 == 0:         # i.e. Even numbers
      even.append(elem[i])

print(odds)
print(even)
```

```python
# Double all numbers in a list

def double_list(numbers):
    new_n = []
    for x in numbers:
      new_n.append(2 * x)

    return new_n

res = double_list([-1, 4, -3, 1, 5, 2, -8, -2, 0])
print(res)
```

Tokenization with NLTK

```python
import nltk # Natural language toolkit
nltk.download('punkt')                                              # use this to download the punkt resource once
data = "Sommaren är bäst i Sverige när bäst är först"
tokens = nltk.word_tokenize(data)
fd = nltk.FreqDist(tokens)
print(tokens)
print(fd)
print(fd.most_common(3))
```

Text Preprocessing with the Pandas library

```python
import pandas as pd

def preprocess_pandas(data):
    data['Sentence'] = data['Sentence'].str.lower()                                                                     # make all lowercase
    data['Sentence'] = data['Sentence'].replace('[a-zA-Z0-9-_.]+@[a-zA-Z0-9-_.]+', '', regex=True)                      # remove emails
    data['Sentence'] = data['Sentence'].replace('((25[0-5]|2[0-4][0-9]|[01]?[0-9][0-9]?)(\.|$)){4}', '', regex=True)    # remove IP address
    data['Sentence'] = data['Sentence'].str.replace('[^\w\s]','')                                                       # remove special characters
    data['Sentence'] = data['Sentence'].replace('\d', '', regex=True)                                                   # remove numbers
    return data

lst = ['Det är mycket bra att laga mat','Swenglish is fun to speak! :)']
df = pd.DataFrame (lst, columns = ['Sentence'])
data = preprocess_pandas(df)
print(data)
```

```python
#!pip install -q kagglehub spacy
#!python -m spacy download en_core_web_sm

import pandas as pd
from pathlib import Path
import kagglehub
import spacy

# download the IMDB dataset -> CSV
dataset_dir = Path(kagglehub.dataset_download(
    "lakshmi25npathi/imdb-dataset-of-50k-movie-reviews"
))
csv_file = next(dataset_dir.rglob("*.csv"))
print(f"Dataset file: {csv_file}")

imdb = pd.read_csv(csv_file)
print(imdb.shape)
imdb.head()
```

```python
# Use spaCy for tokenization and part-of-speech tagging
nlp = spacy.load("en_core_web_sm")
first_20_reviews = imdb["review"].iloc[:20]
documents = list(nlp.pipe(first_20_reviews))

# spaCy tokenization
tokenized_reviews = [
    [token.text for token in document]
    for document in documents
]

# pos tagging
pos_tagged_reviews = [
    [(token.text, token.pos_, token.tag_) for token in document]
    for document in documents
]

print(f"Number of reviews processed: {len(documents)}")
print(f"Token count for review 1: {len(tokenized_reviews[0])}")
```

```python
# display all 20 reviews
summary = pd.DataFrame({
    "review_number": range(1, 21),
    "token_count": [len(tokens) for tokens in tokenized_reviews],
})
display(summary)

for review_number, pos_tags in enumerate(pos_tagged_reviews, start=1):
    print(f"\nReview {review_number}: ")
    print(pos_tags[:20])  # first 20 tagged tokens
    # full data is stored in pos_tagged_reviews
```
